#### Production Caching Architecture

Our LLMOps library implements sophisticated caching at multiple levels:

**Embedding Caching:**
The process of embedding is typically very time consuming and expensive:

1. Send text to OpenAI API endpoint
2. Wait for processing  
3. Receive response
4. Pay for API call

This occurs *every single time* a document gets converted into a vector representation.

**Our Caching Solution:**
1. Check local cache for previously computed embeddings
2. If found: Return cached vector (instant, free)
3. If not found: Call OpenAI API, store result in cache
4. Return vector representation

**LLM Response Caching:**
Similarly, we cache LLM responses to avoid redundant API calls for identical prompts.

**Benefits:**
- ⚡ Faster response times (cache hits are instant)
- 💰 Reduced API costs (no duplicate calls)  
- 🔄 Consistent results for identical inputs
- 📈 Better scalability

Our ProductionRAGChain automatically handles all this caching behind the scenes!

##### ❓ Question #1: Production Caching Analysis

What are some limitations you can see with this caching approach? When is this most/least useful for production systems? 

Consider:
- **Memory vs Disk caching trade-offs**
- **Cache invalidation strategies** 
- **Concurrent access patterns**
- **Cache size management**
- **Cold start scenarios**

> NOTE: There is no single correct answer here! Discuss the trade-offs with your group.

##### ✅ Answer:
Current caching approach weaknesses:

#### **1. Memory vs Disk Caching Trade-offs**

**Memory Caching Limitations:**
- **Memory Constraints**: In-memory caches are limited by available RAM and don't persist across application restarts
- **Scalability Issues**: Memory usage grows linearly with cache size, potentially causing OOM errors
- **Multi-instance Problems**: Each application instance maintains separate caches, leading to redundant storage and cache misses across instances

**Disk Caching Limitations:**
- **I/O Overhead**: Disk reads/writes introduce latency, potentially negating cache benefits for frequently accessed items
- **Storage Management**: Requires disk space management and cleanup strategies
- **Consistency Issues**: File-based caches can become corrupted or inconsistent across concurrent processes

#### **2. Cache Invalidation Strategies**

**Current System Weaknesses:**
- **No TTL (Time-To-Live)**: Cached embeddings and responses never expire, potentially serving stale data indefinitely
- **No Content-Based Invalidation**: When the source document changes, cached embeddings become outdated but aren't automatically invalidated
- **No Version Management**: No mechanism to handle different versions of the same document or embedding model updates

**Production Implications:**
- **Data Staleness**: Users may receive outdated information even after document updates
- **Inconsistent Results**: Different users might get different answers based on cache state
- **Model Drift**: When embedding models are updated, old cached embeddings become incompatible

#### **3. Concurrent Access Patterns**

**Race Conditions:**
- **Cache Stampede**: Multiple concurrent requests for the same uncached item can overwhelm the API
- **Write Conflicts**: Simultaneous cache writes can lead to data corruption or lost updates
- **Inconsistent Reads**: Cache updates might be visible to some processes but not others

**Performance Degradation:**
- **Lock Contention**: Thread-safe caching mechanisms can become bottlenecks under high concurrency
- **Cache Thrashing**: High-frequency cache updates can degrade performance more than no caching

#### **4. Cache Size Management**

**Memory Pressure:**
- **Unbounded Growth**: Without LRU or other eviction policies, caches grow indefinitely
- **Memory Leaks**: Long-running applications may consume excessive memory for rarely-used cached items
- **GC Pressure**: Large in-memory caches can impact garbage collection performance

**Storage Limitations:**
- **Disk Space Exhaustion**: File-based caches can fill available disk space
- **No Intelligent Eviction**: Lack of usage-based cleanup strategies

#### **5. Cold Start Scenarios**

**Initial Performance Impact:**
- **Cache Warming**: New deployments or cache clears result in degraded performance until caches rebuild
- **Uneven Load Distribution**: Some application instances may have "warmer" caches than others
- **User Experience Inconsistency**: First users experience slower responses while building cache state

### **When This Caching Approach is MOST Useful**

- **Static or Rarely-Changing Documents**: When document content changes infrequently
- **Repetitive Query Patterns**: When users frequently ask similar questions
- **High API Costs**: When embedding/LLM API costs are a primary concern
- **Predictable Load Patterns**: When traffic patterns allow for effective cache warming
- **Development/Testing**: For rapid iteration without API cost concerns

### **When This Caching Approach is LEAST Useful**

- **Dynamic Content**: When documents change frequently or need real-time accuracy
- **Unique Query Patterns**: When each query is novel and unlikely to repeat
- **High Concurrency**: When many simultaneous users create cache contention
- **Strict Consistency Requirements**: When data must always be current and accurate
- **Resource-Constrained Environments**: When memory or storage limitations exist
- **Multi-Tenant Systems**: When cache pollution between tenants is a concern


##### 🏗️ Activity #1: Cache Performance Testing

Create a simple experiment that tests our production caching system:

1. **Test embedding cache performance**: Try embedding the same text multiple times
2. **Test LLM cache performance**: Ask the same question multiple times

````python
# 1. Test embedding cache performance: Try embedding the same text multiple times
import time

print("1️⃣ TESTING EMBEDDING CACHE PERFORMANCE")
print("=" * 50)

# Test text for embedding
test_text = "Federal student loan repayment options"

try:
    # First call - should be slower (cache miss)
    print(f"\n🔄 First embedding call (cache miss):")
    start_time = time.time()
    embedding1 = rag_chain.get_retriever().vectorstore.embeddings.embed_query(test_text)
    first_embedding_time = time.time() - start_time
    print(f"⏱️ Time taken: {first_embedding_time:.3f} seconds")
    print(f"📏 Embedding length: {len(embedding1)}")
    
    # Second call - should be faster (cache hit)
    print(f"\n⚡ Second embedding call (cache hit):")
    start_time = time.time()
    embedding2 = rag_chain.get_retriever().vectorstore.embeddings.embed_query(test_text)
    second_embedding_time = time.time() - start_time
    print(f"⏱️ Time taken: {second_embedding_time:.3f} seconds")
    print(f"📏 Embedding length: {len(embedding2)}")
    
    # Third call - should also be cache hit
    print(f"\n⚡ Third embedding call (cache hit):")
    start_time = time.time()
    embedding3 = rag_chain.get_retriever().vectorstore.embeddings.embed_query(test_text)
    third_embedding_time = time.time() - start_time
    print(f"⏱️ Time taken: {third_embedding_time:.3f} seconds")
    
    # Calculate speedup
    if second_embedding_time > 0:
        speedup = first_embedding_time / second_embedding_time
        print(f"\n🚀 Embedding cache speedup: {speedup:.1f}x faster!")
    
except Exception as e:
    print(f"❌ Error testing embedding cache: {e}")

# 2. Test LLM cache performance: Ask the same question multiple times
print("\n\n2️⃣ TESTING LLM CACHE PERFORMANCE")
print("=" * 50)

test_question = "What types of loans are covered in this document?"

try:
    # First call - should be slower (cache miss)
    print(f"\n🔄 First LLM call (cache miss):")
    start_time = time.time()
    response1 = rag_chain.invoke(test_question)
    first_llm_time = time.time() - start_time
    print(f"Response: {response1.content[:100]}...")
    print(f"⏱️ Time taken: {first_llm_time:.2f} seconds")
    
    # Second call - should be faster (cache hit)
    print(f"\n⚡ Second LLM call (cache hit):")
    start_time = time.time()
    response2 = rag_chain.invoke(test_question)
    second_llm_time = time.time() - start_time
    print(f"Response: {response2.content[:100]}...")
    print(f"⏱️ Time taken: {second_llm_time:.2f} seconds")
    
    # Third call - should also be cache hit
    print(f"\n⚡ Third LLM call (cache hit):")
    start_time = time.time()
    response3 = rag_chain.invoke(test_question)
    third_llm_time = time.time() - start_time
    print(f"Response: {response3.content[:100]}...")
    print(f"⏱️ Time taken: {third_llm_time:.2f} seconds")
    
    # Calculate speedup
    if second_llm_time > 0:
        speedup = first_llm_time / second_llm_time
        print(f"\n🚀 LLM cache speedup: {speedup:.1f}x faster!")
    
except Exception as e:
    print(f"❌ Error testing LLM cache: {e}")

# 3. Measure cache hit rates: Compare first call vs subsequent calls
print("\n\n3️⃣ MEASURING CACHE HIT RATES")
print("=" * 50)

# Test multiple different questions to measure hit rates
questions = [
    "What is the purpose of this loan program?",
    "Who is eligible for these loans?", 
    "What is the purpose of this loan program?",  # Repeat for cache test
    "What are the repayment terms?",
    "Who is eligible for these loans?",  # Repeat for cache test
]

cache_hits = 0
total_requests = len(questions)
response_times = []

print("Testing cache hit rates with multiple questions:")

for i, question in enumerate(questions):
    print(f"\nRequest {i+1}: '{question[:40]}...'")
    
    start_time = time.time()
    try:
        response = rag_chain.invoke(question)
        response_time = time.time() - start_time
        response_times.append(response_time)
        
        # Heuristic: responses under 1 second likely indicate cache hits
        if response_time < 1.0:
            cache_hits += 1
            print(f"  ⚡ CACHE HIT - {response_time:.3f}s")
        else:
            print(f"  🔄 CACHE MISS - {response_time:.3f}s")
            
    except Exception as e:
        print(f"  ❌ Error: {e}")

# Calculate and display cache performance metrics
hit_rate = (cache_hits / total_requests) * 100
avg_time = sum(response_times) / len(response_times) if response_times else 0
fastest_time = min(response_times) if response_times else 0
slowest_time = max(response_times) if response_times else 0

print(f"\n📊 CACHE PERFORMANCE SUMMARY:")
print(f"  • Total requests: {total_requests}")
print(f"  • Cache hits: {cache_hits}")
print(f"  • Cache hit rate: {hit_rate:.1f}%")
print(f"  • Average response time: {avg_time:.3f}s")
print(f"  • Fastest response: {fastest_time:.3f}s")
print(f"  • Slowest response: {slowest_time:.3f}s")

print(f"\n✅ Cache performance testing complete!")
````

## Task 4: Guardrails Integration for Production Safety

Now we'll integrate **Guardrails AI** into our production system to ensure our agents operate safely and within acceptable boundaries. Guardrails provide essential safety layers for production LLM applications by validating inputs, outputs, and behaviors.

### 🛡️ What are Guardrails?

Guardrails are specialized validation systems that help "catch" when LLM interactions go outside desired parameters. They operate both **pre-generation** (input validation) and **post-generation** (output validation) to ensure safe, compliant, and on-topic responses.

**Key Categories:**
- **Topic Restriction**: Ensure conversations stay on-topic
- **PII Protection**: Detect and redact sensitive information  
- **Content Moderation**: Filter inappropriate language/content
- **Factuality Checks**: Validate responses against source material
- **Jailbreak Detection**: Prevent adversarial prompt attacks
- **Competitor Monitoring**: Avoid mentioning competitors

### Production Benefits of Guardrails

**🏢 Enterprise Requirements:**
- **Compliance**: Meet regulatory requirements for data protection
- **Brand Safety**: Maintain consistent, appropriate communication tone
- **Risk Mitigation**: Reduce liability from inappropriate AI responses
- **Quality Assurance**: Ensure factual accuracy and relevance

**⚡ Technical Advantages:**
- **Layered Defense**: Multiple validation stages for robust protection
- **Selective Enforcement**: Different guards for different use cases
- **Performance Optimization**: Fast validation without sacrificing accuracy
- **Integration Ready**: Works seamlessly with LangGraph agent workflows

````python
# Import Guardrails components for our production system
print("Setting up Guardrails for production safety...")

try:
    from guardrails.hub import (
        RestrictToTopic,
        DetectJailbreak, 
        CompetitorCheck,
        LlmRagEvaluator,
        HallucinationPrompt,
        ProfanityFree,
        GuardrailsPII
    )
    from guardrails import Guard
    print("✓ Guardrails imports successful!")
    guardrails_available = True
    
except ImportError as e:
    print(f"⚠ Guardrails not available: {e}")
    print("Please follow the setup instructions in the README")
    guardrails_available = False
````

### Demonstrating Core Guardrails

Let's explore the key Guardrails that we'll integrate into our production agent system:

````python
if guardrails_available:
    print("🛡️ Setting up production Guardrails...")
    
    # 1. Topic Restriction Guard - Keep conversations focused on student loans
    topic_guard = Guard().use(
        RestrictToTopic(
            valid_topics=["student loans", "financial aid", "education financing", "loan repayment"],
            invalid_topics=["investment advice", "crypto", "gambling", "politics"],
            disable_classifier=True,
            disable_llm=False,
            on_fail="exception"
        )
    )
    print("✓ Topic restriction guard configured")
    
    # 2. Jailbreak Detection Guard - Prevent adversarial attacks
    jailbreak_guard = Guard().use(DetectJailbreak())
    print("✓ Jailbreak detection guard configured")
    
    # 3. PII Protection Guard - Protect sensitive information
    pii_guard = Guard().use(
        GuardrailsPII(
            entities=["CREDIT_CARD", "SSN", "PHONE_NUMBER", "EMAIL_ADDRESS"], 
            on_fail="fix"
        )
    )
    print("✓ PII protection guard configured")
    
    # 4. Content Moderation Guard - Keep responses professional
    profanity_guard = Guard().use(
        ProfanityFree(threshold=0.8, validation_method="sentence", on_fail="exception")
    )
    print("✓ Content moderation guard configured")
    
    # 5. Factuality Guard - Ensure responses align with context
    factuality_guard = Guard().use(
        LlmRagEvaluator(
            eval_llm_prompt_generator=HallucinationPrompt(prompt_name="hallucination_judge_llm"),
            llm_evaluator_fail_response="hallucinated",
            llm_evaluator_pass_response="factual", 
            llm_callable="gpt-4.1-mini",
            on_fail="exception",
            on="prompt"
        )
    )
    print("✓ Factuality guard configured")
    
    print("\\n🎯 All Guardrails configured for production use!")
    
else:
    print("⚠ Skipping Guardrails setup - not available")
````
